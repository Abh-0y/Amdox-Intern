# **DIFFERENT MODEL IMPLEMENTATION**

```python
#Training Dataset Distribution

train['prognosis'].value_counts()
```

The training dataset is perfectly balanced, meaning each disease category has exactly 120 samples. This ensures that no class is overrepresented or underrepresented, which helps the model learn all diseases equally without bias toward any specific class.

```python
#Train Test Split

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
X = train.drop('prognosis', axis=1)
y = train['prognosis']
```

```python
# Ecoding prognonsis column
le = LabelEncoder()
le.fit(y)
Y = le.transform(y)

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.3, random_state=42)
```

```python
# Import machine learning classification models

from sklearn.svm import LinearSVC
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import  confusion_matrix, accuracy_score, precision_score, recall_score
```

```python
# Create a dictionary of different classification models
models = {
    'SVC': LinearSVC(),
    'MNB': MultinomialNB(),
    'LR': LogisticRegression(penalty='l2'),
    'RFC': RandomForestClassifier(n_estimators=100, random_state=42),
    'GBC': GradientBoostingClassifier(n_estimators=100, random_state=42),
    'KNN': KNeighborsClassifier(n_neighbors=5)
}
```

```python
import warnings
warnings.filterwarnings("ignore")

def evaluate_model(models, x_train=x_train, y_train=y_train, x_test=x_test, y_test=y_test):
  accuracy = []
  precision = []
  recall = []

  for model_name, model in models.items():

    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)
    acc = accuracy_score(y_test, y_pred)
    pre = precision_score(y_test, y_pred, average='weighted')
    rec = recall_score(y_test, y_pred, average='weighted')

    accuracy.append(acc)
    precision.append(pre)
    recall.append(rec)

  return pd.DataFrame({'Model': list(models.keys()), 'Accuracy': accuracy, 'Precision': precision, 'Recall': recall})
```

```python
model_performance = evaluate_model(models)
```

```python
model_performance
```

The model evaluation results show that all the applied algorithms—`SVC`, `Multinomial Naive Bayes (MNB)`, `Logistic Regression (LR)`, `Random Forest Classifier (RFC)`, `Gradient Boosting Classifier (GBC)`, and `K-Nearest Neighbors (KNN)`—achieved perfect performance with 100% `accuracy`, `precision`, and `recall`. This indicates that the dataset is highly structured and easily separable, allowing all models to learn the patterns effectively without misclassification. While this reflects strong model performance, it may also suggest that the dataset is relatively simple or that there is a possibility of overfitting, so further validation on unseen or real-world data is recommended.

# **MODEL FOR PREDICTION**

For model selection, I chose Logistic Regression as the final model. Logistic Regression is simple, fast, and highly interpretable, making it ideal for understanding feature importance. Unlike more complex models, it is less prone to overfitting, easier to implement, and provides clear insights into how each symptom influences the prediction. This makes it an ideal choice for a medical recommendation system, where both accuracy and interpretability are important.

```python
# selecting Logistic Regression
lr = LogisticRegression(penalty='l2')
lr.fit(x_train, y_train)
ypred = lr.predict(x_test)

print("Accuracy of the Logistic Regression: ", accuracy_score(y_test,ypred))
print("Precision of the Logistic Regression: ", precision_score(y_test, ypred, average='weighted'))
print("Recall of the Logistic Regression: ", recall_score(y_test, ypred, average='weighted'))
```

```python
def predict_diseases(patient_symptoms):
    """
    Predict disease and provide related health recommendations.

    Parameters:
    ----------
    patient_symptoms : str
        Comma-separated symptoms entered by the user.
        Example:
        "fever, headache, cough"

    Returns:
    -------
    tuple
        Returns:
        - disease (str)
        - disease_description (str)
        - precautions (list)
        - medications (list)
        - diets (list)
        - workouts (list)

    """

    # Convert symptoms to lowercase
    patient_symptoms = patient_symptoms.lower()

    # Split symptoms and remove extra spaces
    patient_symptoms = [symptom.strip() for symptom in patient_symptoms.split(",")]

    # Create input vector
    input_vector = np.zeros(len(symptoms_dict))

    # Encode symptoms into vector
    for item in patient_symptoms:
        if item in symptoms_dict:
            input_vector[symptoms_dict[item]] = 1
        else:
            print(f"Warning: Symptom '{item}' not found in symptoms_dict. Skipping.")

    # Predict disease
    disease = diseases_list[lr.predict([input_vector])[0]]

    print("================= Predicted Disease =================")
    print(disease)

    # Fetch disease description
    disease_description = desc[desc['Disease'] == disease]['Description']
    disease_description = (
        " ".join([str(w) for w in disease_description.values])
        if not disease_description.empty
        else "No description found."
    )

    # Fetch precautions
    precautions_list = precautions[
        precautions['Disease'] == disease
    ][['Precaution_1', 'Precaution_2', 'Precaution_3', 'Precaution_4']]

    precautions_list = [col for col in precautions_list.values]

    # Fetch medications
    medications_list = medications[
        medications['Disease'] == disease
    ]['Medication']

    medications_list = [med for med in medications_list.values]

    # Fetch diet recommendations
    diet_list = diets[diets['Disease'] == disease]['Diet']
    diet_list = [diet for diet in diet_list.values]

    # Fetch workout recommendations
    workout_list = workout[workout['disease'] == disease]['workout']
    workout_list = [wrk for wrk in workout_list.values]

    # Return all details
    return (
        disease,
        disease_description,
        precautions_list,
        medications_list,
        diet_list,
        workout_list
    )
```
